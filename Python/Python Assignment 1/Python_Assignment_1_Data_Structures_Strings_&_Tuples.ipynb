###**1. String Concatenation:**

Write a Python program that takes two strings, i.e., string 1 "Hello" and string 2 “get
name” as input from the user, and concatenates them together. Display the
concatenated string as the output.

**Sample Output:**

Enter your Name: Zara

Hello Zara

Now, concatenate string 3, "Welcome to Python programming,” to the existing string
and display the output string.

**Sample Output:**

Hello Zara, welcome to Python programming

```python

# 1. String Concatenation

# Get name from user
name = input("Enter your Name: ")

# Concatenate String 1 and String 2
string1 = "Hello "
string2 = name

result = string1 + string2
print(result)

# Add the welcome message
string3 = ", welcome to Python programming"

final_result = result + string3

print(final_result)
```

####**2. String Slicing and Indexing:**

>
Write a Python program using the above concatenated string as input and perform
the following tasks:

a. Print the first character of the string.

b. Print the last character of the string.

c. Print the first 5 characters of the string.

d. Print the last 11 characters of the string.

e. Print the string in reverse.

f. Use slicing and print the word “Python” from the existing string.

```python

# 2. String Slicing and Indexing

# Using the concatenated string from Question 1
full_string = final_result

# a. Print the first character
print("First character:", full_string[0])

# b. Print the last character
print("Last character:", full_string[-1])

# c. Print the first 5 characters
print("First 5 characters:", full_string[:5])

# d. Print the last 11 characters
print("Last 11 characters:", full_string[-11:])

# e. Print the string in reverse
print("Reverse string:", full_string[::-1])

# f. Extract the word Python using slicing
start = full_string.index("Python")
print("Extracted word:", full_string[start:start+6])
```

###**3. String Methods:**
Write a Python program that takes a string, strM = “Python beginner tutorial," and
performs the following tasks:

a. Convert the sentence to uppercase.

b. Convert the sentence to lowercase.

c. Use Capitalize and return the sentence to the original input form.

d. Count the total number of occurrences of character ‘t’ in the string.

e. Replace all occurrences of “Python” with “Data Analytics” in the input string
strM = “Python beginner tutorial.”

```python

# 3. String Methods

# Given input string
strM = "Python beginner tutorial."

# a. Convert the sentence to uppercase
print("Uppercase:", strM.upper())

# b. Convert the sentence to lowercase
print("Lowercase:", strM.lower())

# c. Capitalize the sentence
print("Capitalize:", strM.capitalize())

# d. Count occurrences of character 't'
print("Count of 't':", strM.count('t'))

# e. Replace Python with Data Analytics
print("After replacement:", strM.replace("Python", "Data Analytics"))
```

###**Tuples (Creation, Modification and Access) :**

Create the 1st tuple with values -> (10, 20, 30) and the 2nd tuple with values -> (40, 50, 60):


a. Concatenate the two tuples and store it in “t_combine”

b. Repeat the elements of “t_combine” 3 times

c. Access the 3rd element from “t_combine”

d. Access the first three elements from “t_combine”

e. Access the last three elements from “t_combine”

```python

# 4. Tuples (Creation, Modification and Access)

# Create the first tuple
tuple1 = (10, 20, 30)

# Create the second tuple
tuple2 = (40, 50, 60)

# a. Concatenate the two tuples
t_combine = tuple1 + tuple2
print("Combined Tuple:", t_combine)

# b. Repeat the elements 3 times
print("Repeated Tuple:", t_combine * 3)

# c. Access the 3rd element
print("Third Element:", t_combine[2])

# d. Access the first three elements
print("First Three Elements:", t_combine[:3])

# e. Access the last three elements
print("Last Three Elements:", t_combine[-3:])
```
